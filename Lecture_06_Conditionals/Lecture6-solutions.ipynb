- #### Test if the number satisfies $10 < n < 15$

```python
import random 

lo = 1
hi = 20

random_int = random.randint(lo, hi)
print(random_int)

# SOLUTION

if 1 <= random_int < 5:
    print("number lies in the interval [1, 5)")
elif random_int <= 10: # Q: Why don't we need to check if 5 <= random_int ??
    print("number lies in the interval [5, 10]")
elif random_int < 15: # Q: Why don't we need to check if 10 < random_int ??
    print("number lies in the interval (10, 15)")
else:
    print("number is greater than or equal to 15")
```

#### **7)** Run the cell below. How do you explain the output you see?

```python
list_1 = [0, 0, 0, 0, 0]
list_2 = [5, 50, 75, 100, -0.01]
print(list_1 <= list_2)

# Explanation: This is very similar to alphabetical ordering

# If we put the words "Zebra" and "Apple" in alphabetical order, we
# would put Apple before Zebra, because A comes before Z in the alphabet.
# The endings "pple" and "ebra" do not need to be considered.

# Lists work in a similar way

# Above, since list_1[0] = 0, list_2[0] = 5, and "0 <= 5" is True, 
# Python declares that the statement list_1 <= list_2 is True

# So even though list_1[4] <= list_2[4] is False, 
# comparing the entire lists evaluates to True
```

```python
import random

# Each element is itself a list in the form [name, month, day, year]
birthdays = [
    ["Abed", "March", 23, 2007],
    ["Britta", "July", 14, 1998],
    ["Jeff", "December", 8, 1990],
    ["Shirley", "January", 19, 1990],
    ["Annie", "September", 27, 2007],
    ["Troy", "June", 5, 2007],
    ["Pierce", "December", 30, 1962]]
```

<font size = "5">
Birth season
<br>

- The cell below grabs a random person from `birthdays` 
- Write code that pulls out their month.
- Using the `in` keyword together with `if` / `elif` / `else`, decide which season that month belongs to and store the answer in a variable called `season`.
    - Winter: December, January, February
    - Spring: March, April, May
    - Summer: June, July, August
    - Fall: September, October, November
- Print the result. Try running the cell a few times to make sure it works properly.

```python
person = random.choice(birthdays)
# Alternatively, we could have randomly selected an index and then use that to select a person from birthdays:
# random_index = random.randint(0, len(birthdays)-1)
# person = birthdays[random_index] 
#     
month  = person[1]

if month in ["December", "January", "February"]:
    season = "Winter"
elif month in ["March", "April","May"]:
    season = "Spring"
elif month in ["June", "July", "August"]:
    season = "Summer"
else:
    season = "Fall"


print(f"{person[0]} was born in {month}, which is in {season}.")
```
