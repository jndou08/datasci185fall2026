#### **(1)** The array `temps` below contains the daily high temperatures (in °F) for one week, starting on Sunday and ending on Saturday.

<font size = "4">

- Use negative indexing to print Saturday's temperature
- Use a slice to print the temperatures for Monday through Friday
- Try using a slice with a negative index to print the temperatures for Friday and Saturday

```python
import numpy as np

temps = np.array([71, 68, 74, 80, 77, 65, 69])

# SOLUTION

saturday_temp = temps[-1]
print("Saturday's temperature:", saturday_temp, "degrees (F)")

weekday_temps = temps[1:6]
print("Weekday temperatures:", weekday_temps, "degrees (F)")

weekday_temps_v2 = temps[1:-1]
print("Weekday temperatures:", weekday_temps_v2, "degrees (F)")

# -2 means we start at element 5
# Leave out the second number to go all the way to the end
fri_sat_temp = temps[-2:7]

print("End of week temperatures:", fri_sat_temp, "degrees (F)")

# Leave out the second number to go all the way to the end
fri_sat_temp_v2 = temps[-2:]
print("End of week temperatures:", fri_sat_temp_v2, "degrees (F)")
```

#### (3) In Lecture 4 we saw the list method `.count()`, which counts how many times an item appears in a list. Rewrite the favorite colors example from Section II so that it uses `.count()` instead of `.index()`.

<font size = "4">

*Hint:* loop over the unique colors, instead of over all 15 responses.

```python
favorite_colors = ["red", "red", "blue", "pink", "green", "orange", "orange", "yellow",
                    "red", "green", "red", "blue", "blue", "yellow", "green"]


# Convert favorite_colors to set, eliminating duplicates.
# Then immediately convert it back to a list

unique_colors = list(set(favorite_colors))

# Make an empty list
count = []

for color in unique_colors:
    color_count = favorite_colors.count(color)
    count.append(color_count)

print(unique_colors)
print(count)
```
