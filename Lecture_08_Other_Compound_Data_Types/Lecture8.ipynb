## <span style="color:darkblue"> I. Tuples </span>

<font size = "4">
We'll make a list of 4 elements, then create an equivalent tuple


```python
list_of_numbers = [5, 3, 2, 0]

tuple_of_numbers = (5, 3, 2, 0)

print(list_of_numbers)
print(tuple_of_numbers)
print()
print(type(list_of_numbers))
print(type(tuple_of_numbers))
```

## <span style="color:darkblue; font-size:30px">II. Sets</span>

<font size = "3">

A **set** is an *unordered* collection of *unique* items, ~~just~~ kind of like a set in mathematics.
- Sets are created with curly braces `{ }`
- A set **never contains duplicates**.
- A set has **no order**.
- Sets are **mutable**, like lists. We can add and remove items.

```python
list_of_colors = ["red", "blue", "red", "green", "blue", "red"]

set_of_colors = {"red", "blue", "red", "green", "blue", "red"}

print(list_of_colors)
print(set_of_colors)
print()
print(type(list_of_colors))
print(type(set_of_colors))
```

The duplicates are gone! Also notice that the set might not print in the order we typed it. 


Just like `np.array()` converts a list into an array, `set()` converts a list into a set. This is a quick way to find the **unique** items in a list.

```python
# results of rolling a six-sided die 8 times
rolls = [3, 6, 1, 3, 3, 6, 2, 1]

unique_rolls = set(rolls)
print(rolls)
print(unique_rolls)
print(len(unique_rolls), "different numbers were rolled")
```

```python
# sets of whole numbers are NOT always printed in increasing order
print(list({10, 3, 1}))
```

```python
# Since a set has no order, there is no "first" item. This gives an error!
print(set_of_colors[0])
```

### Checking membership with `in`

<font size = "4">

Creating an empty set

### Set operations

<font size = "4">

Sets come with the same operations you may have seen in math class (think of a Venn diagram with two circles, `a` and `b`):

- **Union** ($a \cup b$), written `a | b`: items in `a` **or** `b` (or both)
- **Intersection** ($a \cap b$), written `a & b`: items in **both** `a` **and** `b`
- **Difference** ($a \setminus b$), written `a - b`: items in `a` but **not** in `b`

**Note:** use the symbols `|` and `&`, not the words `or` and `and`. The words will not give an error, but they will not give you the union or intersection either!

```python
# the days your professor is free for office hours
prof_free = {"Monday", "Thursday", "Friday"}

# the days you are free
my_free = {"Tuesday", "Thursday", "Friday", "Saturday"}

# Intersection: days when BOTH of us are free
print(prof_free & my_free)

# Union: days when AT LEAST ONE of us is free
print(prof_free | my_free)

# Difference: days when the professor is free, but I am NOT
print(prof_free - my_free)
```

```python
# the order matters for difference!
# days when I am free, but the professor is NOT
print(my_free - prof_free)
```

### Example: counting favorite colors

<font size = "4">

Suppose we poll 15 people and ask their favorite color. We simply collect all 15 responses, and store them as strings inside a list. Duplicate answers correspond to duplicate entries of the list. Collecting the data like this is convenient for using `plt.hist` which creates a histogram.

```python
import matplotlib.pyplot as plt

favorite_colors = ["red", "red", "blue", "pink", "green", "orange", "orange", "yellow",
                    "red", "green", "red", "blue", "blue", "yellow", "green"]

plt.hist(x = favorite_colors)
plt.show()
```

The `.index()` method for lists will be useful for demonstration

```python
demo_list = ["first", "second", "third"]

print(demo_list.index("second"))
```

<font size = "4">

For discrete or categorical data such as this, a histogram is not really an appropriate choice (notice how orange and yellow are "stuck together").

Instead we should do a bar plot. But `plt.bar` requires a list with counts of each category, which is not the way we have our data stored.

We can build a list of counts as follows. First, let's understand what we're trying to do:

```python
favorite_colors = ["red", "red", "blue", "pink", "green", "orange", "orange", "yellow",
                    "red", "green", "red", "blue", "blue", "yellow", "green"]

# Want a list where each color appears only *once*
unique_colors = ["red", "blue", "pink", "green", "orange", "yellow"]

# Red appears 4 times, blue appears 3 times, pink appears 1 time, etc.
counts = [4, 3, 1, 3, 2, 2]

plt.bar(unique_colors, counts)

plt.show()
```

<font size = "4">

First, a solution **without** using sets, as proposed by a brave student:

```python
favorite_colors = ["red", "red", "blue", "pink", "green", "orange", "orange", "yellow",
                    "red", "green", "red", "blue", "blue", "yellow", "green"]


unique_colors = []
count = []

for color in favorite_colors:
    if not color in unique_colors:
        unique_colors.append(color)
        count.append(1)
    else:
        count[unique_colors.index(color)] += 1

print(unique_colors)
print(count)
```

<font size = "4">

This solution uses `+=`. This is a shorter way to increase the value of a variable. Here's a simple example demonstrating its use:

```python
x = 5
x = x + 1
print(x)

print("*********")

x = 5
x += 1 # same as x = x + 1
print(x)
```

<font size = "4">

Here is a solution that uses a set that allows us to define the variable `unique_colors` in a single line:

```python
favorite_colors = ["red", "red", "blue", "pink", "green", "orange", "orange", "yellow",
                    "red", "green", "red", "blue", "blue", "yellow", "green"]


# Convert favorite_colors to set, eliminating duplicates.
# Then immediately convert it back to a list

unique_colors = list(set(favorite_colors))

# Make a list full of zeros, with the same length as unique colors.
count = [0] * len(unique_colors)

for color in favorite_colors:
    color_idx = unique_colors.index(color)
    count[color_idx] = count[color_idx] + 1

    # could have had the 2nd indented line be this instead:
    # count[color_idx] += 1

print(unique_colors)
print(count)
```

### Basic Slicing

We can select a range of values using the colon operator `:`.

```python
print(zero2ten)
```

```python
# Extract entries 2,3,4,5
print(zero2ten[2])
print(zero2ten[3])
print(zero2ten[4])
print(zero2ten[5])
```

```python
# This grabs entries 2, 3, 4, but NOT 5 (end is not inclusive)

slice_ = zero2ten[2:5]
print(slice_)
```

```python
# Another example

three2twelve = np.arange(3, 13)
print(three2twelve)
print(three2twelve[2:10])
```

<font size = "4">

Add a third number to take steps larger than 1

```python
print(three2twelve)
print(three2twelve[2:10:2])
```

<font size = "4">

Watch out! Creating a slice does not make a copy of the data!

```python
print(zero2ten)
slice_ = zero2ten[2:5]
print(slice_)
```

```python
# Change 0th element of slice_
slice_[0] = 20
print(slice_)
```

<font size = "4">

But it changes the original array too!

```python
print(zero2ten)
```

<font size = "4">

Use `.copy` to make the slice independent from the original array.

```python
# lets return zero2ten to its previous values.

zero2ten = np.arange(11)
print(zero2ten)

slice_ = zero2ten[2:5].copy()

slice_[0] = 20

print(zero2ten) # doesn't get changed.
```

<font size = "4">

Here's another example where changing "slice_" doesn't change "zero2ten". Understanding the subtleties of when arrays affect each other takes some getting used to:

```python
zero2ten = np.arange(11)

slice_ = zero2ten[2:5] # DON'T use .copy()

zero2ten = np.arange(11) # redefine zero2ten - "decouples" the two variables

slice_[0] = 20

print(zero2ten) # This is just 0, 1, 2, ..., 10. No "20" value is present
```

You can perform element wise operations on slices just like you do with arrays.

```python
zero2ten = np.arange(1, 11)

slice_ = zero2ten[2:5]
print(slice_)

slice_ = slice_ + np.ones(3)*10
print(slice_)
```

```python
# notice that once again, the cell above didn't change the values in zero2ten
# (overwriting slice_ "decouples" it from the variable zero2ten)

print(zero2ten)
```
