## <span style="color:red"> I. Dictionaries </span>

<font size = "5">

- Dictionaries are another Python structure that can hold multiple objects - like lists or NumPy arrays
- Lists and arrays are indexed by the *positions* 0, 1, 2, ...
- Dictionaries are indexed by **keys** that we choose ourselves - for example, people's names

<font size = "5">

Consider the following example:

- We are keeping track of the **lab skills scores** for a small study group of 7 students.
- We decide to use Python lists, which are indexed by the integers 0, 1, 2, ...
- We need two lists: one for the student names, one for their scores.

```python
student_roster = ["Abed", "Britta", "Jeff", "Shirley", "Annie", "Troy", "Pierce"]
skill_scores_list = [3.24, 1.51, 0.51, 6.72, 6.72, 3.66, 0.58]
```

<font size = "5">

- Suppose Jeff attends a lab safety workshop and we want to increase his score to 2.51.

So, if we want to change Jeff's score to 2.51, we could do:

```python
# Jeff is the third student on the list
jeff_index = 2
skill_scores_list[jeff_index] = 2.51 # update corresponding element

print(skill_scores_list)
```

<font size = "5">
However, manually counting to find the location of a student's name would be impractical for a list that is much larger than 7 students. It is also inefficient computationally, and we need to keep track of two variables.

A **dictionary** can be indexed with an integer or a string. We can directly use the students' names.

- We create a dictionary with curly braces `{ }`
- Each entry has the form `key : value`, and entries are separated by commas

```python
skill_scores = {"Abed" : 3.24, "Britta" : 1.51, "Jeff" : 0.51, "Shirley" : 6.72, "Annie" : 6.72, "Troy" : 3.66, "Pierce" : 0.58}

print(skill_scores["Jeff"])
```

<font size = "5">

We can change Jeff's score easily:

```python
skill_scores["Jeff"] = 2.51

print(skill_scores)
```

<font size = "5">

The indices for a dictionary are the **keys**, and the elements of the dictionary are the **values**

```python
print(skill_scores.keys())
print(skill_scores.values())
```

<font size = "4">

We can loop over a dictionary by looping over its keys

```python
for key in skill_scores.keys():
    name = key
    score = skill_scores[name]
    print("Lab skills for", name, ":", score)
```

<font size = "4">

If you try to just loop over the dictionary itself, Python will loop over the keys automatically:

```python
for key in skill_scores:
    print(key)
```

<font size = "5">

- `.keys()` and `.values()` *look* like lists, but they are not (check the `type` below!)
- We can convert them into real lists with `list( )`
- This matters when we want to use list tools on them - for example `random.shuffle` from last class

```python
name_list = list(skill_scores.keys())

print(type(skill_scores.keys()))
print(type(name_list))
print(name_list)

# the same works for the values
score_list = list(skill_scores.values())
print(score_list)
```

```python
# Since name_list is a real list, we can shuffle it into a random order.
# Run this cell a few times!

random.shuffle(name_list)
print(name_list)

# Note that shuffling name_list does NOT change the dictionary itself
print(skill_scores)
```

<font size = "5">

Dictionaries can have strings and integers as keys. The values can be almost any object, including lists (or other dictionaries)

```python
example_dict = {0: "DSCI-160", 
                "favorite movie" : "The Good, the Bad, the Weird",
                12 : 35.2,
                "my_list" : [-8, 0.2, 9.1],
                "roster_from_above" : student_roster, 
                "dict_from_above": skill_scores}

# keys
print(example_dict.keys())
print()

# values
print(example_dict[0])
print(example_dict["favorite movie"])
print(example_dict[12])
print(example_dict["my_list"])
print(example_dict["roster_from_above"])
print(example_dict["dict_from_above"])
```

## <span style="color:red"> II. Dictionaries inside dictionaries </span>

<font size = "5">

In Lectures 6 and 7 we stored each person as a list, `[name, month, day, year]`, and we had to *remember* that the month sits in position 1.

With a dictionary, every piece of information gets a **label**:

```python
abed = {"month": "March", "day": 23, "year": 2007, "lab_skills": 3.24}

print(abed["month"])
print(abed["lab_skills"])
```

<font size = "5">

We can go one step further: a dictionary whose **values are themselves dictionaries**.

- The *outer* keys are the students' names
- Each *inner* dictionary holds everything we know about one student

```python
study_group = {
    "Abed":   {"month": "March", "day": 23, "year": 2007, "lab_skills": 3.24},
    "Britta": {"month": "July",  "day": 14, "year": 1998, "lab_skills": 1.51},
}

print(study_group["Abed"])
```

<font size = "5">

To reach a single piece of information, we do a **two-step lookup**, reading left to right:

- `study_group["Abed"]` is itself a dictionary...
- ...so we can ask *that* dictionary for `["lab_skills"]`

Compare this with `birthdays[2][1]` from Lecture 6: same idea, but now with names instead of positions.

```python
print(study_group["Abed"]["lab_skills"])
print(study_group["Britta"]["month"])
```

## <span style="color:red"> III. Making small DataFrames with dictionaries </span>

<font size = "5">

You can easily make small pandas DataFrames using dictionaries.

- The keys will be the column names of the DataFrame.
- For each key, define the value as a list, which will be the entries of each column.
- **Each list must have the same length.** Otherwise, you will get an error if you try to convert it to a DataFrame

Below, we'll make a small DataFrame containing info for 5 U.S. states:

```python
states = ["Wisconsin", "Georgia", "Delaware", "Illinois", "Oregon"]

# Wisconsin was 30th state admitted to the U.S., Georgia was 4th state admitted etc.
state_id = [30, 4, 1, 21, 33]

capitals = ["Madison", "Atlanta", "Dover", "Springfield", "Salem"]

# year admitted to the U.S.
admit_date = ["1848", "1788", "1787", "1818", "1859"]

# number of electoral votes for each state
electoral_votes = [10, 16, 3, 19, 8]

state_dict = {"state_id": state_id, "state_name" : states, "capital_city" : capitals, 
                "year_admitted" : admit_date, "electoral_college_votes" : electoral_votes}


df_states = pd.DataFrame(state_dict)

display(df_states)
```

<font size = "5">

We'll make another small DataFrame with different info for the same 5 states.

```python
big_city = ["Milwaukee", "Atlanta", "Wilmington", "Chicago", "Portland"]

state_dict_2 = {"state_id" : state_id, "largest_city" : big_city, "year_admitted" : admit_date}

df_cities = pd.DataFrame(state_dict_2)

display(df_cities)
```
