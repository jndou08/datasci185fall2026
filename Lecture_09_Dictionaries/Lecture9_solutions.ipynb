**1.** The cell below defines quiz scores for 5 students.

- Print Duncan's score.
- Chang complains and gets 2 extra points. Update his score (use his *current* score - don't just type the new number).
- Using a `for` loop over the keys, print one line per student in the form `Quiz score for Chang : 8.5`

```python
quiz_scores = {"Chang" : 6.5, "Duncan" : 9.0, "Craig" : 4.5, "Vicki" : 7.5, "Garrett" : 5.0}

# your code below
print(quiz_scores["Duncan"])

quiz_scores["Chang"] = quiz_scores["Chang"]+2

for k in quiz_scores.keys():
    print(f"Quiz score for {k}: {quiz_scores[k]}")
```

**2.** Find the *best* student in `quiz_scores` using a **running maximum**, like in last class's exercises:

- Before the loop, create `best_score = 0` and `best_name = "nobody"`.
- Loop over the keys. Whenever you meet a score higher than `best_score`, update *both* variables.
- Print the winner's name and score.
- Check your answer for the score with `np.max` (you will need `list( )` and `.values( )`).

```python
import numpy as np
best_score = 0
best_name = "nobody"

# your code below

for k in quiz_scores.keys():
    if quiz_scores[k] > best_score:
        best_score = quiz_scores[k]
        best_name = k

print(f'Best quiz score was {best_score} from {best_name}')
# checking with np.max
print(np.max(list(quiz_scores.values())))
```
