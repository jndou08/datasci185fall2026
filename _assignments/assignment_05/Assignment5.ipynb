<font size = "3">

We have a **single** loop variable `i` that takes the values 0, 1, and 2. But the value of "i" wasn't directly needed...we only used the loop variable to define `current_title` and `current_rating`, which is what we actually used in the body of the loop.

In some sense, we had **two** loop variables ("current_title" and "current_rating"), although we explicitly defined only one.

The `zip` function actually allows us to explicitly define two loop variables. This is demonstrated below:

```python
good_movies = [] # create empty list

cut_off = 8.0 # minimum rating for "good movie"

movie_titles = ["The Good, the Bad, the Weird", "Citizen Kane" , "Spider-Man: Across the Spider-Verse" ]
movie_ratings = [7.2, 8.2, 8.5]

for current_title, current_rating in zip(movie_titles, movie_ratings):
    if current_rating >= cut_off:
        good_movies.append(current_title)

print(good_movies)
```

Below is the function "quadratic_formula" from Lecture 10, along with 3 calls to the function with different input parameters. If we call the function with `a=1`, `b=2`, `c=5`, we receive a warning message and the output tuple's two elements are both `np.nan` (nan means "not a number").

- Determine why this behavior happens in the third call to the function, but not the first two. (No written explanation is necessary, you'll show you know why when you complete the second bullet point)

- Change the function so that when a similar situation is encountered as the third call, the function returns a single string "No real roots" and no warning message is displayed.


```python
import numpy as np

# quadratic_formula from Lecture 10
def quadratic_formula(a, b, c):
    discriminant = b**2 - 4*a*c

    if discriminant < 0:
        return "No real roots"

    else:
        sqrt_d = np.sqrt(discriminant)

        numer_plus = -b + sqrt_d
        numer_minus = -b - sqrt_d

        denom = 2*a

        x_plus = numer_plus/denom
        x_minus = numer_minus/denom
        return x_plus, x_minus

print(quadratic_formula(5, 0, -5))
print(quadratic_formula(1, 2, -8))
print(quadratic_formula(1, 2, 5))
```

**(d)** More function revision

<font size = "3">

The quadratic formula is intended to solve equations of the form $$ax^2 + bx + c = 0$$

But what if $a = 0$? Then we are trying to solve $bx + c = 0$ and the quadratic formula is not appropriate.

The function below is written to dissuade the user from passing `a=0` into the function:

```python
def quadratic_formula_safe(a, b, c):
    if a == 0:
        print("*** Warning: the leading coefficient must be non-zero")
        x_plus = None
        x_minus = None
        return x_plus, x_minus

    else:
        discriminant = b**2 - 4*a*c
        sqrt_d = np.sqrt(discriminant)

        numer_plus = -b + sqrt_d
        numer_minus = -b - sqrt_d

        denom = 2*a

        x_plus = numer_plus/denom
        x_minus = numer_minus/denom
        return x_plus, x_minus

print(quadratic_formula_safe(5, 0, -5))
print(quadratic_formula_safe(1, 2, -8))
print(quadratic_formula_safe(0, 1, 2))
```

The cell below creates a DataFrame containing student information from a (small) course. The professor decides to curve every grade using the formula $$\text{new grade} = 10\sqrt{\text{old grade}}$$


- Write a `lambda` function that takes in the old grade as input, and outputs the new (curved) grade. The new grade should be rounded to **one decimal place**.


- Use the `.apply()` method with your lambda function to compute the curved grades, rounded to 1 decimal place. This should return a pandas.Series object with the new grades. Change the `.name` attribute of this object to the string "curved_grade".

- Write a `lambda` function that returns `True` if a grade is at least 55 and `False` otherwise.

- Use your function with the `.apply()` method to count how many students pass **before** and **after** the curve, without writing a loop. (Hint: what is `True + True`?)

```python
import numpy as np
import pandas as pd 

student_info = pd.DataFrame({"name": ["Chang", "Duncan", "Craig", "Vicki", "Garrett"],
                             "id_no": ["A-490", "K-158", "Z-240", "C-111", "E-150"],
                             "major": ["Biology", "Data Science", "Physics", "English", "Biology"],
                             "numeric_grade": [85, 90, 45, 75, 50]})


# Your code here
grade_curver = lambda old_grade: round(10 * np.sqrt(old_grade), 1)
pass_checker = lambda grade: grade >= 55

curved_grades = student_info["numeric_grade"].apply(grade_curver)
curved_grades.name = "curved_grade"
print(curved_grades)

pass_nocurve = student_info["numeric_grade"].apply(pass_checker).sum()
pass_curve = curved_grades.apply(pass_checker).sum()

print(f"The number of students that passed before the curve was {pass_nocurve}.")
print(f"The number of students that passed after the curve was {pass_curve}.")


```
