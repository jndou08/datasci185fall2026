The code below generates a random temperature and also randomly chooses the temperature scale (Celsius or Fahrenheit). Define `feels` so that the following phrases are printed, depending on the randomly generated variables.

- "It feels cold" if the scale is Fahrenheit and temperature is below 60
- "It feels cold" if the scale is Celsius and the temperature is below 15.
- "It feels hot" if the scale is Fahrenheit and the temperature is above 80
- "It feels hot" if the scale is Celsius and the temperature is above 26, but less than or equal to 48
- "It feels like I am dying" if the scale is Celsius and the temperature is above 48.
- "It feels just right" otherwise

Can you solve this problem without using the `and` operator at all?

```python
import random

# SOLUTION 1

temperature = random.randint(0, 110)
temp_scale = random.choice(["Fahrenheit", "Celsius"])

if temp_scale == "Fahrenheit":
    if temperature < 60:
        feels = "cold"
    elif temperature > 80:
        feels = "hot"
    else:
        feels = "just right"
else:
    if temperature < 15:
        feels = "cold"
    elif temperature > 48:
        feels = "like I am dying"
    elif temperature > 26: # Q: Why don't we need to check if temperature <= 48 here ?
        feels = "hot"
    else:
        feels = "just right"


print(temperature)
print(temp_scale)

print(f"It feels {feels}")
```

**Question (C)** Stopping a for loop

<font size = "3">

Add up the values in `receipts` but stop if the sum exceeds 90.

Print the string "stopped at receipt # {receipt_number}" after the `for` loop ends with the number of the last reciept you counted (assume the reciepts are numbered starting at zero)

```python
import numpy as np

### NOTE: multiplying by 15 doesn't make the sum exceed 90 very often.
### So we will multiply by 30 instead

# receipts = np.random.uniform(size = 10)*15

receipts = np.random.uniform(size = 10)*30 
print(receipts)

receipt_number = -1
my_sum = 0

for value in receipts:
    receipt_number = receipt_number + 1
    my_sum = my_sum + value
    if my_sum > 90:
        break

print(f"stopped at receipt # {receipt_number}")
```

**Question (D)** Simulate a sequence of dice rolls

<font size = "3">

Suppose we want to see how many rolls of a six-sided die it takes to roll a "3" **twice** (in total, not consecutively). Write code that simulates this experiment using `random.choice` to simulate a single dice roll along with a `while` loop. After your code completes, the variable `num_rolls` should be equal to the number of rolls it took to roll the second "3".

**Example:**

If we roll the following numbers (reading left-to-right):

$$ 1, 4, 3, 1, 1, 1, 2, 3$$

then `num_rolls` should have the value 8.



```python
num_rolls = 0

### SOLUTION
import random

dice_faces = [1, 2, 3, 4, 5, 6]

num_threes = 0 # how many threes we have rolled

while num_threes < 2:
    # roll the die
    roll = random.choice(dice_faces)

    # we rolled the die, so num_rolls should be increased by 1
    num_rolls = num_rolls + 1

    # did we roll a three?
    if roll == 3:
        # increase the number of threes we've seen by 1
        num_threes = num_threes + 1

# that's it, print the number of rolls
print("Number of rolls:", num_rolls)


```

- `pop_estimate`- A list containing current population estimates of the countries (in millions of people). They are ordered in the same way as the two tuples. E.g., "pop_estimate[1]" is 23.243, because Taiwan's estimated population is 23,243,000 people.*

1. Use the `.index` method to find where the string "Hellenic Republic" is located inside `formal_names`. Assign this to a variable called `change_index`.

2. Find that country's informal name inside of `informal_names` using the variable you created in step 1. Also find the out-of-date population estimate inside `pop_estimate` using the same variable.

3. Print the message "Old population estimate of [informal name here] is [population value here] million". You should not explicitly type the strings corresponding to the informal name or the population value.

4. Change the appropriate element of `pop_estimate` to 9.267, using the variable you created in step 1.

5. Print the message "Current population estimate of [informal name here] is [population value here] million". You should not explicitly type the strings corresponding to the informal name or the population value.


\* Data taken from [britannica.com](https://www.britannica.com/topic/list-of-countries-dependencies-and-territories-by-population).

```python
formal_names = ("United Mexican States", "Republic of China", "People's Republic of China",
                "Romania", "Lebanese Republic", "State of Kuwait",
                "Democratic Socialist Republic of Sri Lanka", "Kingdom of Sweden",
                "Hellenic Republic", "People's Democratic Republic of Algeria",
                "Socialist Republic of Viet Nam", "Commonwealth of Australia")

informal_names = ("Mexico", "Taiwan", "China",
                  "Romania", "Lebanon", "Kuwait",
                  "Sri Lanka", "Sweden", 
                  "Greece", "Algeria",
                  "Vietnam", "Australia")

pop_estimate = [135.674, 23.243, 1407.698,
                19.089, 5.821, 5.149, 
                21.938, 10.616,
                10.432, 48.472,
                103.439, 28.060]


## SOLUTION
change_index = formal_names.index("Hellenic Republic")
name = informal_names[change_index]
print("Old population estimate of", name, "is", pop_estimate[change_index], "million.")
pop_estimate[change_index] = 9.267
print("Current population estimate of", name, "is", pop_estimate[change_index], "million.")
```

**Question (F)** Removing elements from a set

<font size = "3">

In the cell below, a set is randomly chosen between two sets. The set `course_schedule` contains the names of the classes a student is currently registered for. Do the following:

1. Add the course "BIOL 301" to the set.

2. Add code that removes the following courses (if they belong to the set): "MATH 250", "DATASCI 325", and "MATH 346". Note that, depending on which choice of `course_schedule` is selected, none of these classes may appear, one class may appear, or multiple may appear. Your code should successfully remove the correct courses without causing an error, regardless of which schedule is chosen. If the schedule chosen has none of the classes, then no items should be removed.

```python
import random

course_schedule_a = {"DATASCI 185", "DATASCI 210", "MATH 250", "BIOL 223"}
course_schedule_b = {"DATASCI 185", "DATASCI 325", "MATH 315", "BIOL 223"} 
course_schedule_c = {"DATASCI 185", "DATASCI 210", "MATH 346", "BIOL 223"} 
course_schedule_c = {"DATASCI 325", "DATASCI 210", "MATH 346", "BIOL 223"} 
course_schedule_d = {"DATASCI 185", "DATASCI 210", "ENG 205", "BIOL 223"} 
course_schedule_e = {"DATASCI 325", "MATH 346", "MATH 315", "MATH 250"}

schedule_list = [course_schedule_a, course_schedule_b, course_schedule_c, course_schedule_d, course_schedule_e]

course_schedule = random.choice(schedule_list)

print(course_schedule)


### SOLUTION 1 - no looping
course_schedule.add("BIOL 301")

if "MATH 250" in course_schedule:
    course_schedule.remove("MATH 250")
if "DATASCI 325" in course_schedule:
    course_schedule.remove("DATASCI 325")
if "MATH 346" in course_schedule:
    course_schedule.remove("MATH 346")

print(course_schedule)
```

```python
import random

course_schedule_a = {"DATASCI 185", "DATASCI 210", "MATH 250", "BIOL 223"}
course_schedule_b = {"DATASCI 185", "DATASCI 325", "MATH 315", "BIOL 223"} 
course_schedule_c = {"DATASCI 185", "DATASCI 210", "MATH 346", "BIOL 223"} 
course_schedule_c = {"DATASCI 325", "DATASCI 210", "MATH 346", "BIOL 223"} 
course_schedule_d = {"DATASCI 185", "DATASCI 210", "ENG 205", "BIOL 223"} 
course_schedule_e = {"DATASCI 325", "MATH 346", "MATH 315", "MATH 250"}

schedule_list = [course_schedule_a, course_schedule_b, course_schedule_c, course_schedule_d, course_schedule_e]

course_schedule = random.choice(schedule_list)

print(course_schedule)


### SOLUTION 2 - use a for loop
course_schedule.add("BIOL 301")

remove_courses = ["MATH 250", "DATASCI 325", "MATH 346"]

for course in remove_courses:
    if course in course_schedule:
        course_schedule.remove(course)


print(course_schedule)
```
